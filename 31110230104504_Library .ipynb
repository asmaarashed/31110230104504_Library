```python
import pandas as pd
import sqlite3
```

# Task 1: Data Gathering and Combination

```python
conn = sqlite3.connect('Library Database.db')

Books=pd.read_sql_query("SELECT * FROM Books",conn)
members=pd.read_sql_query("SELECT * FROM members",conn)
checkouts=pd.read_sql_query("SELECT * FROM checkouts",conn)
```

Q1_ How much is each member borrowing?

```python
q = """
SELECT
    members.member_id,
    members.first_name,
    members.last_name,
    COUNT(checkouts.checkout_id) AS total_checkouts
FROM members

LEFT JOIN checkouts ON members.member_id = checkouts.member_id
GROUP BY  members.member_id,members.first_name,members.last_name
"""

pd.read_sql_query(q, conn)
```

Q2_Which books match a chosen author pattern?

```python
q="""
select book_id,title,author
from books

where author like 'A%'
"""
pd.read_sql_query(q,conn)
```

Q3_What are the most popular books?


```python
q="""
SELECT
    Books.book_id,
    Books.title,
    COUNT(checkouts.checkout_id) AS total_checkouts
FROM Books

JOIN checkouts ON Books.book_id = checkouts.book_id
GROUP BY Books.book_id , Books.title

order by total_checkouts desc
limit 5
"""
pd.read_sql_query(q,conn)
```

Q4_Who are the most active readers?

```python
q="""
SELECT
    members.member_id,
    members.first_name,
    members.last_name,
    COUNT(checkouts.checkout_id) AS total_checkouts
FROM members

JOIN checkouts ON members.member_id = checkouts.member_id
GROUP BY members.member_id,members.first_name,members.last_name

order by total_checkouts desc
limit 10
"""
pd.read_sql_query(q,conn)
```

Q5_What does a neighborhood's activity look like further back in time?

```python
q="""
SELECT
    checkouts.checkout_date AS checkout_date,
    members.neighborhood,
    checkouts.checkout_id,
    checkouts.return_date,
    checkouts.member_id,
    checkouts.book_id,
    members.first_name,
    members.last_name
FROM members

JOIN checkouts ON members.member_id = checkouts.member_id

where neighborhood = 'Maadi'

order by checkouts.checkout_date DESC
limit 10
offset 10
"""
pd.read_sql_query(q,conn)
```

```python
book_catalog=pd.read_json('Book Catalog.json')
book_catalog.head()
```

```python
kickoff=pd.read_html('Reading_Kickoff_Signups.html')
kickoff=kickoff[0]

kickoff.columns=['member_id','book_id','checkout_date']
kickoff.head()
```

Stage 1

```python
member_counts = checkouts['member_id'].value_counts().reset_index(name='total_books')
member_counts.columns = ['member_id', 'total_books']
```

```python
stage1 = pd.merge(checkouts, members, on='member_id', how='left')
stage1 = pd.merge(stage1, member_counts, on='member_id', how='left')
```

Stage 2

```python
full_books = pd.merge(Books, book_catalog, on='book_id', how='left')
stage2 = pd.merge(stage1, full_books, on='book_id', how='left')
```

Stage 3

```python
kickoff = pd.merge(kickoff, members, on='member_id', how='left')

kickoff = pd.merge(kickoff, full_books, on='book_id', how='left')

kickoff['checkout_id'] = pd.NA
kickoff['return_date'] = pd.NA
kickoff['total_books'] = pd.NA # it will be calculate later

kickoff= kickoff[stage2.columns]

final_df = pd.concat([stage2, kickoff], ignore_index=True)

final_df['total_books'] = (final_df.groupby('member_id')['member_id'].transform('size'))
```
